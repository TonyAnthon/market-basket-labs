```python
# Create sample transaction data
data = {'TransactionID': [1, 1, 2, 2, 2, 3, 3, 4, 4, 4],
        'Item': ['Milk', 'Bread', 'Bread', 'Butter', 'Milk', 'Milk', 'Diapers', 'Bread', 'Butter', 'Diapers']}
df_trans = pd.DataFrame(data)

# Create a pivot table (One-hot encoding via crosstab)
basket = pd.crosstab(df_trans['TransactionID'], df_trans['Item'])
basket
```

```python
frequent_itemsets_comprehensive = apriori(basket, min_support=0.5, use_colnames=True)
print("Frequent Itemsets:")
frequent_itemsets_comprehensive
```

```python
rules_comprehensive = association_rules(frequent_itemsets_comprehensive, metric="confidence", min_threshold=0.6)
print("Association Rules:")
rules_comprehensive
```
